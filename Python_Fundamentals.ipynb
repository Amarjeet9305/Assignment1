## 1. Data Type & Variable

A **variable** is a name that refers to a value stored in memory.
Python is dynamically typed — you don't need to declare a type; it is inferred from the value.

Common built-in data types:
- `int` — whole numbers
- `float` — decimal numbers
- `str` — text
- `bool` — True/False
- `list` — ordered, mutable collection
- `tuple` — ordered, immutable collection
- `dict` — key-value pairs
- `set` — unordered collection of unique items

```python
# Step 1: Creating variables of different data types
name = "Alice"          # str
age = 21                # int
height = 5.6             # float
is_student = True        # bool
subjects = ["Math", "Physics", "CS"]   # list
coordinates = (10.0, 20.0)             # tuple
student_info = {"name": "Alice", "age": 21}  # dict
unique_ids = {101, 102, 103}           # set

# Step 2: Printing the values
print(name, age, height, is_student)
print(subjects)
print(coordinates)
print(student_info)
print(unique_ids)
```

```python
# Step 3: Checking data types with type()
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
print(type(subjects))
print(type(coordinates))
print(type(student_info))
print(type(unique_ids))
```

```python
# Step 4: Type conversion (casting)
x = "25"
y = int(x)        # string -> int
z = float(y)       # int -> float
w = str(z)          # float -> string

print(x, type(x))
print(y, type(y))
print(z, type(z))
print(w, type(w))
```

## 2. Operator and Expression

An **operator** performs an operation on values (operands).
An **expression** is a combination of variables, values, and operators that evaluates to a result.

Types of operators:
- Arithmetic: `+ - * / // % **`
- Comparison (Relational): `== != > < >= <=`
- Logical: `and or not`
- Assignment: `= += -= *= /=`
- Membership: `in`, `not in`
- Identity: `is`, `is not`

```python
# Step 1: Arithmetic operators
a = 15
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

```python
# Step 2: Comparison operators
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
# Step 3: Logical operators
p = True
q = False

print(p and q)
print(p or q)
print(not p)
```

```python
# Step 4: Assignment operators
c = 10
c += 5   # c = c + 5
print(c)
c -= 3   # c = c - 3
print(c)
c *= 2   # c = c * 2
print(c)
c //= 4  # c = c // 4
print(c)
```

```python
# Step 5: Membership and Identity operators
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)
print("grape" not in fruits)

x = [1, 2, 3]
y = [1, 2, 3]
z = x

print(x is z)       # same object
print(x is y)       # different objects, same values
print(x == y)       # value comparison
```

## 3. Conditional Statement

Conditional statements let a program make decisions using `if`, `elif`, and `else`.
The code block runs only when the condition evaluates to `True`.

```python
# Step 1: Simple if statement
marks = 85

if marks >= 40:
    print("You passed!")
```

```python
# Step 2: if-else statement
marks = 30

if marks >= 40:
    print("You passed!")
else:
    print("You failed. Try again.")
```

```python
# Step 3: if-elif-else ladder (grading system)
marks = 78

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
```

```python
# Step 4: Nested if statements
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Entry denied: must be 18+")
```

## 4. Loop and Iterations

Loops let you repeat a block of code multiple times.
Python has two main loop types:
- `for` loop — iterates over a sequence (list, string, range, etc.)
- `while` loop — repeats as long as a condition is `True`

Loop control statements: `break`, `continue`, `pass`

```python
# Step 1: for loop with range()
for i in range(1, 6):
    print("Iteration:", i)
```

```python
# Step 2: for loop over a list
fruits = ["apple", "banana", "mango", "grapes"]

for fruit in fruits:
    print("Fruit:", fruit)
```

```python
# Step 3: for loop over a string (character iteration)
for ch in "Python":
    print(ch)
```

```python
# Step 4: while loop
count = 1

while count <= 5:
    print("Count is:", count)
    count += 1
```

```python
# Step 5: break statement - stop loop when condition is met
for num in range(1, 10):
    if num == 5:
        break
    print(num)
```

```python
# Step 6: continue statement - skip current iteration
for num in range(1, 10):
    if num % 2 == 0:
        continue
    print("Odd number:", num)
```

```python
# Step 7: nested loops - multiplication table
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i*j}")
    print("---")
```

## 5. Function

A **function** is a reusable block of code that performs a specific task.
Defined using the `def` keyword. Functions can take parameters and return values.

```python
# Step 1: Defining and calling a simple function
def greet():
    print("Hello, welcome to Python!")

greet()
```

```python
# Step 2: Function with parameters
def greet_user(name):
    print(f"Hello, {name}! Welcome aboard.")

greet_user("Ravi")
greet_user("Sneha")
```

```python
# Step 3: Function with return value
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 25)
print("Sum:", result)
```

```python
# Step 4: Function with default parameter values
def power(base, exponent=2):
    return base ** exponent

print(power(5))        # uses default exponent = 2
print(power(2, 3))     # overrides default
```

```python
# Step 5: Function with variable number of arguments (*args)
def total_sum(*numbers):
    return sum(numbers)

print(total_sum(1, 2, 3))
print(total_sum(10, 20, 30, 40))
```

```python
# Step 6: Lambda function (anonymous function)
square = lambda x: x * x
print(square(6))

# Using lambda with map()
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print(squared)
```

## 6. Slicing

Slicing extracts a portion (sub-sequence) of a sequence (string, list, tuple)
using the syntax: `sequence[start:stop:step]`
- `start` — index to begin (inclusive), default 0
- `stop` — index to end (exclusive), default end of sequence
- `step` — how many indices to skip, default 1

```python
# Step 1: Slicing a string
text = "Python Programming"

print(text[0:6])     # 'Python'
print(text[7:])      # 'Programming'
print(text[:6])      # 'Python'
print(text[-11:])    # last 11 characters
```

```python
# Step 2: Slicing with a step
text = "Python Programming"

print(text[::2])     # every 2nd character
print(text[::-1])    # reversed string
```

```python
# Step 3: Slicing a list
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[1:4])     # [20, 30, 40]
print(numbers[:3])      # [10, 20, 30]
print(numbers[3:])      # [40, 50, 60, 70]
print(numbers[::2])     # [10, 30, 50, 70]
print(numbers[::-1])    # reversed list
```

```python
# Step 4: Slicing a tuple
coordinates = (1, 2, 3, 4, 5, 6, 7, 8)

print(coordinates[2:6])
print(coordinates[-3:])
```

```python
# Step 5: Modifying a list using slicing
numbers = [1, 2, 3, 4, 5]
numbers[1:3] = [20, 30]   # replace elements at index 1 and 2
print(numbers)
```
